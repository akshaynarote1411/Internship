### Task-1 : Predictive Modeling

#### 1. Build a regression model to predict the aggregate rating of a restaurant based on available features. Split the dataset into training and testing sets and evaluate the model's performance using appropriate metrics.

The columns with the unique values are not much useful in the prediction and the column which has name of city, name of restaurants, address, etc are not useful in the prediction so we should drop the unneccessary columns from the dataset.

```python
# Dropping unneccessary features from the dataset 
df1 = df.drop(columns = ["Restaurant ID","Restaurant Name","Country Code","City","Address","Locality","Locality Verbose","Longitude","Latitude","Cuisines","Currency",])
```

```python
# Data Wrangling : 
X = pd.get_dummies(df1.drop("Aggregate rating",axis=1),drop_first = True)   # It will do the encoding of discrete_categorical input columns using dummy or nominal encoding by dropping 1st category columns in dummy tables.
print(X)                                                          # It will do encoding as True or False, it will not convert categorical values into 0's or 1's
print("-"*100)
y = df1["Aggregate rating"]
print(y)

# Splitting the dataset into two parts train_set and test_set
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8,random_state=6)

```

```python
# Creating the linear Regression model to predict Aggregate Rating of a restaurant based on the available features.
     
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,y_train)

# Printing the coeffiecient and intercept for linear regression model
print("coefficient:",model.coef_)
print("intercept:",model.intercept_)
print("-"*100)
```

```python
# Predicting on the identified Linear Regression model
    
ypred_train = model.predict(X_train)
print("ypred_train:",ypred_train)
ypred_test = model.predict(X_test)
print("ypred_test:",ypred_test)
print("-"*100)

# Evaluation by using different metrices 

from sklearn.metrics import mean_absolute_error
MAE_train = mean_absolute_error(y_train,ypred_train)
print("MAE_train:",MAE_train)
MAE_test = mean_absolute_error(y_test,ypred_test)
print("MAE_test:",MAE_test)
print("-"*100)

from sklearn.metrics import mean_squared_error
MSE_train = mean_squared_error(y_train,ypred_train)
print("MSE_train:",MSE_train)
MSE_test = mean_squared_error(y_test,ypred_test)
print("MSE_test:",MSE_test)
print("-"*100)

RMSE_train = np.sqrt(MSE_train)
print("RMSE_train:",RMSE_train)
RMSE_test = np.sqrt(MSE_test)
print("RMSE_test:",RMSE_test)
print("-"*100)
```

```python
# Cheching whether train accuracy = test accuracy or not for the LR model performance   
train_accuracy = model.score(X_train,y_train)
print("train_accuracy:",train_accuracy)
test_accuracy = model.score(X_test,y_test)
print("test_accuracy:",test_accuracy)
print("-"*100)

# Checking whether test accuracy = cross validation score or not for the LR model performance
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model,X,y,cv=5)
print(scores)
print("cross validation score:",scores.mean())
```

#### 2. Experiment with different algorithms (e.g., linear regression, decision trees, random forest) and compare their performance.

```python
# Creating the Decision Tree model to predict Aggregate Rating of a restaurant based on the available features.

from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor(random_state=0)
model.fit(X_train,y_train)


# Predicting on the identified Decision Tree model

ypred_train = model.predict(X_train)
print("ypred_train:",ypred_train)
ypred_test = model.predict(X_test)
print("ypred_test:",ypred_test)
print("-"*100)


# Evaluation by using different metrices

from sklearn.metrics import mean_absolute_error
MAE_train = mean_absolute_error(y_train,ypred_train)
print("MAE_train:",MAE_train)
MAE_test = mean_absolute_error(y_test,ypred_test)
print("MAE_test:",MAE_test)
print("-"*100)

from sklearn.metrics import mean_squared_error
MSE_train = mean_squared_error(y_train,ypred_train)
print("MSE_train:",MSE_train)
MSE_test = mean_squared_error(y_test,ypred_test)
print("MSE_test:",MSE_test)
print("-"*100)

RMSE_train = np.sqrt(MSE_train)
print("RMSE_train:",RMSE_train)
RMSE_test = np.sqrt(MSE_test)
print("RMSE_test:",RMSE_test)
print("-"*100)

train_accuracy = model.score(X_train,y_train)
print("train_accuracy:",train_accuracy)
test_accuracy = model.score(X_test,y_test)
print("test_accuracy:",test_accuracy)

from sklearn.model_selection import cross_val_score
print("cross_val_score:",cross_val_score(model,X,y,cv=5).mean())
```

```python
# Creating the Decision Tree model to predict Aggregate Rating of a restaurant based on the available features.

from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(random_state=0)
model.fit(X_train,y_train)


# Predicting on the identified Decision Tree model

ypred_train = model.predict(X_train)
print("ypred_train:",ypred_train)
ypred_test = model.predict(X_test)
print("ypred_test:",ypred_test)
print("-"*100)


# Evaluation by using different metrices

from sklearn.metrics import mean_absolute_error
MAE_train = mean_absolute_error(y_train,ypred_train)
print("MAE_train:",MAE_train)
MAE_test = mean_absolute_error(y_test,ypred_test)
print("MAE_test:",MAE_test)
print("-"*100)

from sklearn.metrics import mean_squared_error
MSE_train = mean_squared_error(y_train,ypred_train)
print("MSE_train:",MSE_train)
MSE_test = mean_squared_error(y_test,ypred_test)
print("MSE_test:",MSE_test)
print("-"*100)

RMSE_train = np.sqrt(MSE_train)
print("RMSE_train:",RMSE_train)
RMSE_test = np.sqrt(MSE_test)
print("RMSE_test:",RMSE_test)
print("-"*100)

train_accuracy = model.score(X_train,y_train)
print("train_accuracy:",train_accuracy)
test_accuracy = model.score(X_test,y_test)
print("test_accuracy:",test_accuracy)

from sklearn.model_selection import cross_val_score
print("cross_val_score:",cross_val_score(model,X,y,cv=5).mean())
```

### Task-3 : Data Visualization 

#### 1. Create visualizations to represent the distribution of ratings using different charts (histogram, barplot, etc.).

```python
# Plotting the Histogram to show the distribution of ratings.
sns.histplot(df["Aggregate rating"],bins=30,color="red",kde=True)
plt.xlabel("Rating")
plt.ylabel("Count/Frequency")
plt.title("Distribution of Rating by using Histogram ")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="Aggregate rating",data=df,color="red",width=0.8)
plt.xlabel("Rating")
plt.ylabel("Count/Frequency")
plt.title("Distribution of Rating by using Bars ")
plt.show()
```

#### 2. Compare the average ratings of different cuisines or cities using appropriate visualizations.

```python
# Plotting the average rating of restaurants based on the diiferent cuisines.
plt.figure(figsize=(10,5))
sns.barplot(x=df.groupby("Cuisines")["Aggregate rating"].mean().head(30).index,
            y=df.groupby("Cuisines")["Aggregate rating"].mean().head(30).values, 
            data=df,color="green",width=0.4)
plt.xticks(rotation=-90)
plt.ylabel("Rating")
plt.xlabel("Cuisines")
plt.title("Distribution of Average Rating for diiferent Cuisines ")
plt.show()
```

```python
# Displaying the average rating of restaurants based on the different Cities.
plt.figure(figsize=(10,5))
sns.barplot(x=df.groupby("City")["Aggregate rating"].mean().head(30).index,
            y=df.groupby("City")["Aggregate rating"].mean().head(30).values, 
            data=df,color="green",width=0.4)
plt.xticks(rotation=-90)
plt.xlabel("Cities")
plt.ylabel("Rating")
plt.title("Distribution of Average Rating for different Cities")
plt.show()
```
