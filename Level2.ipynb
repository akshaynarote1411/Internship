#### 2. Compare the average ratings of restaurants with table booking and those without.

#### 3. Analyze the availability of online delivery among restaurants with different price ranges.

```python
# Finding the percentage of the resturants which offers online delivery with different price ranges.
Delivery_available_counts = df.groupby("Price range")["Has Online delivery"].value_counts(normalize=True).unstack()*100

# Ploting the graph for the Online Delivery offers by Restaurants with different Price Ranges.
Delivery_available_counts.plot(kind='bar',stacked=True,figsize=(10,7))
plt.title("Online Delivery offers by Restaurants with different Price Ranges.")
plt.xlabel("Price Range")
plt.ylabel("Percentage of Restautants")
plt.show()
```

#### Analysis on the availability of online delivery among restaurants with different price ranges:
- The more resturants offers online delivery on the middle price range.
- There are less restaurants which offers online delivery with high price range compared to other price range.

### Task-3 : Feature Engineering

#### 1. Extract additional features from the existing columns, such as the length of the restaurant name or address.

```python
# Creating a new feature which has values with length of the restaurant name
df["Length_of_name"] = df["Restaurant Name"].str.len()
df["Length_of_name"]
```

```python
# Creating the new feature which has values with address length
df["Length_of_address"] = df["Address"].str.len()
df["Length_of_address"]
```

```python
# Extracting new creating features
df[["Restaurant Name","Length_of_name","Address","Length_of_address"]]
```

#### 2. Create new features like "Has Table Booking" or "Has Online Delivery" by encoding categorical variables.

```python
# Extracting the required features from the old dataset to the new dataset
df1 = df.iloc[:,[12,13]]

# Enconding the categorical values to numerical values
df1.replace({'Yes':1,'No':0},inplace=True)
df1
```

```python
# Creating the new features in the dataset with encoded numerical values
df["Has Table booking_encoded"] = df1["Has Table booking"]
df["Has Online delivery_encoded"] = df1["Has Online delivery"]
df
```

```python
# Extracting the categorical features with the respected encoded new features.
df[["Has Table booking","Has Table booking_encoded","Has Online delivery","Has Online delivery_encoded"]]
```
